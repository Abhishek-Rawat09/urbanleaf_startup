```python
df_sample['Cancelled_order'] = (df_sample['order_status'] == 'Cancelled').astype(int)
df_sample['Processing_order'] = (df_sample['order_status'] == 'Processing').astype(int)
df_sample['Shipped_order'] = (df_sample['order_status'] == 'Shipped').astype(int)
```

```python
# Successful delivered purchases
delivered_df = df[df["order_status"] == "Delivered"].copy()
```

```python
delivered_df
```

```python
delivered_df.groupby('gender')['customer_rating'].mean()
```

```python
delivered_df["customer_rating"] = delivered_df.groupby("gender")["customer_rating"].transform(
    lambda x: x.fillna(round(x.mean(), 1))
)
```

```python
delivered_df.info()
```

```python
delivered_df['order_date'] = pd.to_datetime(delivered_df['order_date'])
delivered_df['delivery_date'] = pd.to_datetime(delivered_df['delivery_date'])
```
